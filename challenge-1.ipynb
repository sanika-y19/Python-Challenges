```python
#Students marks calculator

a = int(input("Enter subject 1 marks: "))
b = int(input("Enter subject 2 marks: "))
c = int(input("Enter subject 3 marks: "))

total = a + b + c
average = total / 3   

print("Total marks: ",total)
print("Average marks: ",average)

if a >= 35 and b >= 35 and c >= 35:
    print("Result: PASS")
else:
    print("Result: FAIL")
    
```
